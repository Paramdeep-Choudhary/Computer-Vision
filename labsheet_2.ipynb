```python
# ============================================================
# EXPERIMENT NO. 2
# CONTRAST ENHANCEMENT AND HISTOGRAM-BASED IMAGE PROCESSING
# USING PYTHON AND OPENCV
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

print("Libraries imported successfully!")


# ------------------------------------------------------------
# 2. UPLOAD IMAGE
# ------------------------------------------------------------

print("Upload a grayscale or color image:")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

print("Image uploaded:", filename)


# ------------------------------------------------------------
# 3. READ IMAGE
# ------------------------------------------------------------

image = cv2.imread(filename)

if image is None:
    raise Exception("Image could not be loaded.")

# Convert to grayscale
gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

print("Image converted to grayscale successfully.")


# ------------------------------------------------------------
# 4. DISPLAY ORIGINAL IMAGE
# ------------------------------------------------------------

plt.figure(figsize=(6, 5))

plt.imshow(
    gray,
    cmap="gray"
)

plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()


# ------------------------------------------------------------
# 5. ORIGINAL IMAGE HISTOGRAM
# ------------------------------------------------------------

plt.figure(figsize=(7, 4))

plt.hist(
    gray.ravel(),
    bins=256,
    range=[0, 256]
)

plt.title("Histogram of Original Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()


# ------------------------------------------------------------
# 6. CONTRAST STRETCHING
# ------------------------------------------------------------

min_val = np.min(gray)
max_val = np.max(gray)

if max_val > min_val:

    contrast_stretched = (
        (gray - min_val)
        * 255
        / (max_val - min_val)
    ).astype(np.uint8)

else:

    contrast_stretched = gray.copy()


# ------------------------------------------------------------
# 7. HISTOGRAM EQUALIZATION
# ------------------------------------------------------------

hist_equalized = cv2.equalizeHist(
    gray
)


# ------------------------------------------------------------
# 8. CLAHE
# ------------------------------------------------------------

clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

clahe_image = clahe.apply(
    gray
)


# ------------------------------------------------------------
# 9. DISPLAY ALL ENHANCED IMAGES
# ------------------------------------------------------------

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)

plt.imshow(
    gray,
    cmap="gray"
)

plt.title("Original")
plt.axis("off")


plt.subplot(2, 2, 2)

plt.imshow(
    contrast_stretched,
    cmap="gray"
)

plt.title("Contrast Stretching")
plt.axis("off")


plt.subplot(2, 2, 3)

plt.imshow(
    hist_equalized,
    cmap="gray"
)

plt.title("Histogram Equalization")
plt.axis("off")


plt.subplot(2, 2, 4)

plt.imshow(
    clahe_image,
    cmap="gray"
)

plt.title("CLAHE")
plt.axis("off")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 10. HISTOGRAM COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)

plt.hist(
    gray.ravel(),
    bins=256,
    range=[0, 256]
)

plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")


plt.subplot(2, 2, 2)

plt.hist(
    contrast_stretched.ravel(),
    bins=256,
    range=[0, 256]
)

plt.title("Contrast Stretching Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")


plt.subplot(2, 2, 3)

plt.hist(
    hist_equalized.ravel(),
    bins=256,
    range=[0, 256]
)

plt.title("Histogram Equalization")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")


plt.subplot(2, 2, 4)

plt.hist(
    clahe_image.ravel(),
    bins=256,
    range=[0, 256]
)

plt.title("CLAHE Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 11. SIDE-BY-SIDE COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(16, 5))

plt.subplot(1, 4, 1)

plt.imshow(
    gray,
    cmap="gray"
)

plt.title("Original")
plt.axis("off")


plt.subplot(1, 4, 2)

plt.imshow(
    contrast_stretched,
    cmap="gray"
)

plt.title("Contrast Stretching")
plt.axis("off")


plt.subplot(1, 4, 3)

plt.imshow(
    hist_equalized,
    cmap="gray"
)

plt.title("Histogram Equalization")
plt.axis("off")


plt.subplot(1, 4, 4)

plt.imshow(
    clahe_image,
    cmap="gray"
)

plt.title("CLAHE")
plt.axis("off")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. SAVE OUTPUT IMAGES
# ------------------------------------------------------------

cv2.imwrite(
    "original_gray.jpg",
    gray
)

cv2.imwrite(
    "contrast_stretched.jpg",
    contrast_stretched
)

cv2.imwrite(
    "histogram_equalized.jpg",
    hist_equalized
)

cv2.imwrite(
    "clahe_output.jpg",
    clahe_image
)

print("\nOutput images saved successfully!")

print("""
============================================================
                    OBSERVATIONS
============================================================

1. The original image was converted into grayscale.

2. The histogram of the original image was generated to
   analyze pixel intensity distribution.

3. Contrast stretching expanded the range of pixel
   intensities and improved overall contrast.

4. Histogram Equalization redistributed pixel intensities
   and enhanced image contrast.

5. CLAHE improved local contrast while limiting excessive
   contrast enhancement.

6. CLAHE generally performs better than standard histogram
   equalization when illumination is uneven.

7. Histogram-based enhancement improves the visibility of
   important features in an image.

8. These techniques are useful in medical imaging,
   surveillance, satellite imaging, and computer vision.

============================================================
              EXPERIMENT COMPLETED SUCCESSFULLY
============================================================
""")
```
