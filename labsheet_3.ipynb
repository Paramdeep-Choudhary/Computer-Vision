```python
# ============================================================
# EXPERIMENT NO. 3
# IMPLEMENTATION AND ANALYSIS OF SPATIAL FILTERING TECHNIQUES
# USING LOW-PASS AND HIGH-PASS FILTERS
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

print("Libraries imported successfully!")


# ------------------------------------------------------------
# 2. UPLOAD IMAGE
# ------------------------------------------------------------

print("Upload an image:")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

print("Image uploaded:", filename)


# ------------------------------------------------------------
# 3. READ IMAGE AND CONVERT TO GRAYSCALE
# ------------------------------------------------------------

image = cv2.imread(filename)

if image is None:
    raise Exception("Image could not be loaded.")

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

print("Image loaded and converted to grayscale.")


# ------------------------------------------------------------
# 4. DISPLAY ORIGINAL IMAGE
# ------------------------------------------------------------

plt.figure(figsize=(6, 5))

plt.imshow(
    gray,
    cmap="gray"
)

plt.title("Original Image")
plt.axis("off")
plt.show()


# ============================================================
# LOW-PASS FILTERS
# ============================================================

# ------------------------------------------------------------
# 5. GAUSSIAN FILTER
# ------------------------------------------------------------

gaussian = cv2.GaussianBlur(
    gray,
    (5, 5),
    0
)

print("Gaussian filtering completed.")


# ------------------------------------------------------------
# 6. MEDIAN FILTER
# ------------------------------------------------------------

median = cv2.medianBlur(
    gray,
    5
)

print("Median filtering completed.")


# ------------------------------------------------------------
# 7. AVERAGE / MEAN FILTER
# ------------------------------------------------------------

average = cv2.blur(
    gray,
    (5, 5)
)

print("Average filtering completed.")


# ============================================================
# HIGH-PASS FILTERS
# ============================================================

# ------------------------------------------------------------
# 8. LAPLACIAN FILTER
# ------------------------------------------------------------

laplacian = cv2.Laplacian(
    gray,
    cv2.CV_64F
)

laplacian = cv2.convertScaleAbs(
    laplacian
)

print("Laplacian filtering completed.")


# ------------------------------------------------------------
# 9. SOBEL HORIZONTAL FILTER
# ------------------------------------------------------------

sobel_x = cv2.Sobel(
    gray,
    cv2.CV_64F,
    1,
    0,
    ksize=3
)

sobel_x = cv2.convertScaleAbs(
    sobel_x
)

print("Sobel X filtering completed.")


# ------------------------------------------------------------
# 10. SOBEL VERTICAL FILTER
# ------------------------------------------------------------

sobel_y = cv2.Sobel(
    gray,
    cv2.CV_64F,
    0,
    1,
    ksize=3
)

sobel_y = cv2.convertScaleAbs(
    sobel_y
)

print("Sobel Y filtering completed.")


# ------------------------------------------------------------
# 11. SOBEL COMBINED
# ------------------------------------------------------------

sobel_combined = cv2.addWeighted(
    sobel_x,
    0.5,
    sobel_y,
    0.5,
    0
)

print("Combined Sobel filtering completed.")


# ============================================================
# 12. DISPLAY LOW-PASS FILTER RESULTS
# ============================================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(
    gray,
    cmap="gray"
)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(
    gaussian,
    cmap="gray"
)
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(
    median,
    cmap="gray"
)
plt.title("Median Filter")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(
    average,
    cmap="gray"
)
plt.title("Average / Mean Filter")
plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 13. DISPLAY HIGH-PASS FILTER RESULTS
# ============================================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(
    gray,
    cmap="gray"
)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(
    laplacian,
    cmap="gray"
)
plt.title("Laplacian Filter")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(
    sobel_x,
    cmap="gray"
)
plt.title("Sobel X - Horizontal Edges")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(
    sobel_y,
    cmap="gray"
)
plt.title("Sobel Y - Vertical Edges")
plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 14. DISPLAY COMBINED SOBEL RESULT
# ============================================================

plt.figure(figsize=(7, 5))

plt.imshow(
    sobel_combined,
    cmap="gray"
)

plt.title("Combined Sobel Edge Detection")
plt.axis("off")

plt.show()


# ============================================================
# 15. SAVE FILTERED IMAGES
# ============================================================

cv2.imwrite(
    "original_gray.jpg",
    gray
)

cv2.imwrite(
    "gaussian_filter.jpg",
    gaussian
)

cv2.imwrite(
    "median_filter.jpg",
    median
)

cv2.imwrite(
    "average_filter.jpg",
    average
)

cv2.imwrite(
    "laplacian_filter.jpg",
    laplacian
)

cv2.imwrite(
    "sobel_x.jpg",
    sobel_x
)

cv2.imwrite(
    "sobel_y.jpg",
    sobel_y
)

cv2.imwrite(
    "sobel_combined.jpg",
    sobel_combined
)

print("\nAll filtered images saved successfully!")


# ============================================================
# 16. FINAL OBSERVATIONS
# ============================================================

print("""
============================================================
                    OBSERVATIONS
============================================================

1. Gaussian filtering smooths the image and reduces noise.

2. Median filtering is effective for removing salt-and-pepper
   noise while preserving edges.

3. Average filtering smooths the image but may blur edges.

4. Laplacian filtering highlights edges and fine details.

5. Sobel X detects intensity changes mainly in the horizontal
   direction.

6. Sobel Y detects intensity changes mainly in the vertical
   direction.

7. Low-pass filters are mainly used for smoothing and noise
   reduction.

8. High-pass filters are mainly used for edge detection and
   detail enhancement.

9. Filtering is an important preprocessing step in computer
   vision.

============================================================
              EXPERIMENT COMPLETED SUCCESSFULLY
============================================================
""")
```
